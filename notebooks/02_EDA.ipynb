```python
# 2. Import Libraries & Load Clean Data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Plot settings
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Load cleaned dataset
df = pd.read_csv("../data/processed/farm_ml_clean_data.csv")

df.head()
```

# 4. Univariate Analysis

```python
# Profit Distribution (Histogram)
plt.figure(figsize=(10,5))
sns.histplot(df['Profit'], bins=30, kde=True)
plt.title("Profit Distribution Across Farms")
plt.xlabel("Profit (INR)")
plt.ylabel("Number of Farms")
plt.tight_layout()
plt.show()
```

### Insight
Most farms earn moderate profits, while only a few farms earn very high profits.

```python
# Revenue Distribution (Histogram)
plt.figure(figsize=(10,5))
sns.histplot(df['Total_Revenue'], bins=30, kde=True, color='green')
plt.title("Revenue Distribution Across Farms")
plt.xlabel("Revenue")
plt.ylabel("Number of Farms")
plt.tight_layout()
plt.show()
```

### Insight

Most farms have similar revenue, but a few farms generate much higher revenue than others.


```python
# Profit by State (Boxplot)
plt.figure(figsize=(12,5))
sns.boxplot(x='State', y='Profit', data=df)
plt.title("Profit Distribution by State")
plt.tight_layout()
plt.show()
```

### Insight

Some states have higher average profits than others, showing that farm performance varies by location.

```python
# Mortality Rate Distribution
plt.figure(figsize=(10,5))
sns.violinplot(y=df["Mortality_Rate"])
plt.title("Mortality Rate Distribution")
plt.xlabel("Mortality Rate")
plt.ylabel("Number of Farms")
plt.tight_layout()
plt.show()
```

### Insight

Most farms have low mortality rates, while a few farms have noticeably higher mortality.

# 5. Bivariate Analysis

```python
# Mortality Rate vs Profit
plt.figure(figsize=(10,5))
sns.scatterplot(x='Mortality_Rate', y='Profit', data=df)
plt.title("Mortality Rate vs Profit")
plt.xlabel("Mortality Rate")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

### Insight

Profit generally decreases as mortality rate increases.

```python
# Revenue vs Feed Cost
plt.figure(figsize=(10,5))
sns.scatterplot(
    x='Total_Feed_Cost',
    y='Total_Revenue',
    data=df
)
plt.title("Revenue vs Feed Cost")
plt.xlabel("Feed Cost (INR)")
plt.ylabel("Revenue (INR)")
plt.tight_layout()
plt.show()
```

### Insight

Farms with higher feed costs usually generate higher revenue.

```python
# Feed Efficiency vs Profit
df["Feed_Efficiency"] = (df["Total_Feed_Used_KG"] /df["Total_Chicks"].replace(0, 1))

plt.figure(figsize=(10,5))
sns.scatterplot(x='Feed_Efficiency', y='Profit', data=df)
plt.title("Feed Efficiency vs Profit")
plt.xlabel("Feed Efficiency")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

### Insight

Better feed efficiency is associated with higher farm profits.

# 6. Correlation Analysis

```python
numeric_df = df.select_dtypes(include="number")
plt.figure(figsize=(12,8))
sns.heatmap(numeric_df.corr(),annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

### Insight

Profit is positively related to revenue and negatively related to mortality rate.

# 7. Business Insights

```python
# Top 10 Revenue Generating Farms
top_revenue = df.sort_values(by='Total_Revenue', ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(x='Farmer_ID', y='Total_Revenue', data=top_revenue)
plt.title("Top 10 Revenue Generating Farms")
plt.tight_layout()
plt.show()
```

### Insight

A small number of farms contribute the highest share of total revenue.

```python
# Top States by Revenue 
state_rev = (df.groupby("State")["Total_Revenue"].sum().sort_values(ascending=False))
sns.barplot(x=state_rev.index,y=state_rev.values)
plt.title("Top 10 State by Revenue")
plt.tight_layout()
plt.show()
```

### Insight

Some states generate much higher total revenue than others.

```python
# Convert Join_Date to datetime
df["Join_Date"] = pd.to_datetime(df["Join_Date"])

# Extract Year-Month
df["Join_Month"] = df["Join_Date"].dt.to_period("M").astype(str)

# Count farms registered each month
monthly = (
    df.groupby("Join_Month")
      .size()
      .reset_index(name="Farm_Count")
)

plt.figure(figsize=(12,5))
sns.lineplot(data=monthly, x="Join_Month", y="Farm_Count", marker="o")
plt.xticks(rotation=90)
plt.title("Monthly Farm Registration Trend")
plt.xlabel("Month")
plt.ylabel("Number of Farms")
plt.tight_layout()
plt.show()
```

### Insight

The number of new farm registrations changes over time, showing periods of higher and lower registration activity.

# EDA Summary & Key Findings

### Key Findings

- Most farms earn moderate profits, while a few farms perform exceptionally well.
- Revenue generally increases with feed cost.
- Higher mortality rates are associated with lower profits.
- Feed efficiency appears to support better profitability.
- Farm performance varies across different states.
- A small number of farms generate a large share of total revenue.
- Farm registrations fluctuate over time.
- These findings will be used in the next notebook to create meaningful features for machine learning models.

Note: Profit is calculated as Total_Revenue - Total_Feed_Cost. Negative profit values indicate that a farm's feed cost exceeded its revenue, representing a financial loss. These values are valid and have been retained for analysis.
